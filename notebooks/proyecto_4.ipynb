````python


# ¿Todos los pedidos con valores ausentes tienen más de 64 productos?
# Si
# Agrupa todos los pedidos con datos ausentes por su ID de pedido.
conteo_productos_por_pedido = order_products.groupby('order_id')['product_id'].count()

# Cuenta el número de 'product_id' en cada pedido y revisa el valor mínimo del conteo.
conteos_nan = conteo_productos_por_pedido[conteo_productos_por_pedido.index.isin(ids_con_na)]

print('Número de pedidos con valores ausentes:', len(conteos_nan))
print('Mínimo de productos en esos pedidos:', conteos_nan.min())
print('Máximo de productos en esos pedidos:', conteos_nan.max())
````

### [C3] Para cada producto, ¿cuál es la proporción de las veces que se pide y que se vuelve a pedir?

````python
proporcion_por_producto = order_products.groupby('product_id')['reordered'].mean()
df_proporciones = proporcion_por_producto.reset_index()
df_merged= df_proporciones.merge(products, on='product_id').reset_index()

df_merged['reordered'].hist(bins=30)
plt.title('Distribución de Proporciones de Reorden')
plt.xlabel('Proporción de Reorden')
plt.ylabel('Número de Productos')

top_productos = df_merged.nlargest(20, 'reordered')
top_productos.plot(x='product_name', y='reordered', kind='barh', figsize=(10, 8))


````

Escribe aquí tus conclusiones 
Se calculó la proporción de reorden promediando la columna reordered para cada producto. Este valor indica qué tan frecuente es que un artículo sea comprado nuevamente por los clientes. Luego se unió con el dataset products para mostrar el ID, nombre y proporción de recompra de cada producto.

### [C5] ¿Cuáles son los 20 principales artículos que las personas ponen primero en sus carritos?

````python

primeros_productos = order_products[order_products['add_to_cart_order'] == 1]

top_20 = (
    primeros_productos
    .groupby('product_id')['order_id']
    .nunique()  # cuenta pedidos únicos
    .nlargest(20)
    .reset_index(name='first_count')
)
df_top20 = top_20.merge(products[['product_id', 'product_name']], on='product_id')

df_top20 = df_top20.sort_values('first_count', ascending=False)[['product_id', 'product_name', 'first_count']]
df_top20.head(20)
````

````python
df_top20.plot(x='product_name', y='first_count', kind='barh')
plt.gca().invert_yaxis()  # para mostrar el más alto arriba
plt.xlabel('Veces que fue el primer artículo')
plt.title('Top 20 productos puestos primero en el carrito')
plt.show()
````
